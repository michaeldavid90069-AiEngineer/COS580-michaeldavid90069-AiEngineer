```python
# --- 1. READING AND SPLITTING THE DATA ---
import pandas as pd
from sklearn.model_selection import train_test_split

# Load the flight prices dataset
df = pd.read_csv('flight_prices.csv')

# Separate features and continuous target
X = df.drop(columns=['price'])
y = df['price']

# Perform an 80/20 train-test split (No stratify for regression)
# Using random_state=42 to ensure reproducibility
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Output the shape of the split datasets to verify
print(f"Training Features Shape: {X_train.shape}")
print(f"Test Features Shape: {X_test.shape}")


# --- 2. LINEAR REGRESSION MODEL ---
from sklearn.linear_model import LinearRegression

# Initialize and fit Linear Regression using scikit-learn
# Hyperparameter change: positive=True (forces positive coefficients)
lr_model = LinearRegression(positive=True)
lr_model.fit(X_train, y_train)
print("\nLinear Regression model successfully fitted.")


# --- 3. K-NEAREST NEIGHBORS (KNN) REGRESSION MODEL ---
from sklearn.neighbors import KNeighborsRegressor

# Initialize and fit KNN Regression
# Hyperparameter changes: 7 neighbors, distance-based weighting, and Manhattan distance (p=1)
knn_model = KNeighborsRegressor(n_neighbors=7, weights='distance', p=1)
knn_model.fit(X_train, y_train)
print("KNN Regression model successfully fitted.")

# NOTE: No scoring metrics are calculated yet, per the instructions!
```
